```python
import pandas as pd

url = "https://storage.googleapis.com/download.tensorflow.org/data/creditcard.csv"
df  = pd.read_csv(url)
df.head()
```

```python
# problemas de classificação desbalanceada

df["Class"].value_counts(normalize=True)
```

```python
# criar variaveis para ajudar o modelo
import numpy as np

df["Amount_log"] =np.log1p(df["Amount"])
```

```python
# transformação da media = 0 e desvio padrão = 0

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df["Amount_scaled"] = scaler.fit_transform(df[["Amount"]])
```

```python
#preparar os dados para treinamento
from sklearn.model_selection import train_test_split

X = df.drop(["Class", "Amount"], axis=1)
y = df["Class"]

X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y,test_size=0.3, random_state=42)
```

```python
# Avaliação do modelo
from sklearn.linear_model import LogisticRegression

model1 = LogisticRegression(max_iter=1000)

model1.fit(X_train, y_train)

y_pred = model1.predict(X_test)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

y_pred_proba = model1.predict_proba(X_test)[:,1]

fpr, tpr, _ = roc_curve(y_test, y_pred_proba)

plt.plot(fpr, tpr)
plt.title("ROC Curve")
plt.xlabel("False posite curve")
plt.ylabel("True positive curve")
plt.show()
```

```python
from sklearn.metrics import precision_recall_curve

precision, recall, _ = precision_recall_curve(y_test, y_pred_proba)

plt.plot(recall, precision)
plt.title("Precision Recall Curve")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.show()
```

```python
# balanceamento dos dados
from imblearn.over_sampling  import SMOTE

smote = SMOTE()
X_res, y_res = smote.fit_resample(X, y)

```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=50,
    max_depth=10,
    class_weight="balanced",
    n_jobs=-1,
    random_state=42
)

rf.fit(X_res, y_res)

y_pred_rf = rf.predict(X_test)

print(classification_report(y_test, y_pred_rf))

```

```python
# organizar o fluxo de processamento
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000))
])

pipeline.fit(X_train, y_train)

y_pred = pipeline.predict(X_test)
```

```python
# criar modelo
model = RandomForestClassifier()

# treinar modelo
model.fit(X_train, y_train)

threshold = 0.3
y_probs = model.predict_proba(X_test)[:,1]

y_pred_custom = (y_probs > threshold).astype(int)

print(classification_report(y_test, y_pred_custom))


```
